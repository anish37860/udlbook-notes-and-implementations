# Chapter 02: SUPERVISED LEARNING

Supervised learning fits a mathematical function to data by adjusting internal parameters such that inputs accurately map to target outputs.

$$
\mathbf{y} = {\bf f}[\mathbf{x}, \boldsymbol\phi] \tag{2.1}
$$

- **Model & Inference**: A supervised model is a mathematical function that maps input ($\mathbf{x_i}$) (e.g., car age and mileage) to output ($\mathbf{y_i}$) (e.g., car value). When a model computes an output from any input (known/unknown), it is called inference. 

- **Parameters** ($\boldsymbol\phi$): Equations contain trainable parameters that define the specific relationship between inputs and outputs from a broader family of possible functions.

- **Training & Generalization**: A learning algorithm trains on a training set, which comprises known input/output pairs, to tune the model's parameters. While the immediate goal is to accurately predict the outputs of the training set using their corresponding inputs, the ultimate goal is generalization to new, unseen inputs.

Consider Eqn $2.1$, It is a family of relation between $\mathbf{x}$ & $\mathbf{y}$ and by tuning $\boldsymbol\phi$, we are determining the particular relation which best map the inputs with outputs. To quantify the error in this mapping, we use a *loss function*, $L$.

- **Loss Function** ($L[\boldsymbol\phi]$): A scalar metric representing how far the predicted outputs deviate from the target training outputs for a certain parameter, $\boldsymbol\phi$. Hence, the final goal is to find a parameter ($\hat{\boldsymbol\phi}$) which minimizes the loss function.
$$
\hat{\boldsymbol\phi} = \mathop{\rm argmin}\limits_{\boldsymbol\phi}\Bigl[L\left[\boldsymbol\phi\right] \Bigr] \tag{2.2}
$$

### 2.1 Linear regression example
Let's consider a model $y = f[x, ϕ]$ that predicts a single output $y$ from a single input $x$, This type of model is known as **1D linear regression model** and it represents a straight line in 2D space.
$$
y = f[x,\boldsymbol\phi] \\
y  =\phi_{0}+\phi_{1}x \tag{2.3}
$$
This model has two parameters $\boldsymbol\phi = [\phi_0, \phi_1]$, where $\phi_0$ and $\phi_1$ is the y-intercept and slope of the line respectively [[Interactive figure for Fig 2.1](https://udlbook.github.io/udlfigures/)].

#### 2.1.1 Loss 
For this model, the training dataset consists of $I$ input/output pairs $\{x_i, y_i\}$. The mismatch is captured by the deviation between the model prediction $y = f[x_i,\boldsymbol\phi]$ (height of the line at $x_i$) and the ground truth outputs $y_i$. 
Hence, we quantify the total mismatch/deviation which we called **training error** or **loss** as the sum of the squares of these deviations for all $I$ training pairs:
$$
L[\boldsymbol\phi] = \sum_{i=1}^{I} \left(f[x_{i}, \boldsymbol\phi]-y_{i}\right)^{2}\nonumber \\
L[\boldsymbol\phi] = \sum_{i=1}^{I} \left(\phi_{0}+\phi_{1}x_i-y_{i}\right)^{2} \tag{2.4}
$$
Since the best parameters minimize this expression, we call this a **least-squares loss**. The squaring operation means that the direction of the deviation (i.e., whether the line is above or below the data) is unimportant. We read the theoretical reasons for this choice in the chapter 5.
As it's a squared, it can only take zero or +ve values, hence, we can say that
$ L[\boldsymbol\phi] \geq 0 $.

As our final goal is to find a parameter ($\hat{\boldsymbol\phi}$) which minimizes the loss function $L[\boldsymbol\phi]$:
$$
\hat{\boldsymbol\phi} = \mathop{\rm argmin}\limits_{\boldsymbol\phi}\Bigl[L[\boldsymbol\phi]\Bigr] \\
\hat{\boldsymbol\phi} = \mathop{\rm argmin}\limits_{\boldsymbol\phi}\left[\sum_{i=1}^{I} \left({f}[x_{i}, \boldsymbol\phi]-y_{i}\right)^{2}\right] \\
\hat{\boldsymbol\phi} = \mathop{\rm argmin}\limits_{\boldsymbol\phi}\left[\sum_{i=1}^{I} \left(\phi_{0}+\phi_{1}x_i-y_{i}\right)^{2}\right] \tag{2.5}
$$
There are two parameters, the y-intercept ($\phi_0$) and slope ($\phi_1$). So we can calculate the loss for every combination of values and visualize the loss function as  surface (Figure 2.3).

#### 2.1.2 Training 
The process of finding parameter values that minimize the loss function is known as *model fitting, training or learning*. To do that we use, **The Gradient Descent Algorithm** which includes these steps [[Interactive figure of Fig 2.4](https://udlbook.github.io/udlfigures/)]:
1. Initialization: Assign random starting values to the parameters.
2. Gradient Measurement: Calculate the slope (gradient) of the loss surface at the current position.
3. Step Downhill: Adjust parameters in the direction of steepest decrease.
4. Convergence: Repeat the cycle until the gradient becomes flat (zero), indicating the minimum loss has been reached.

#### 2.1.2 Testing
Evaluating a trained model's real-world performance requires computing its loss on an independent test dataset. How well prediction accuracy generalizes to this new data depends on data quality (representativeness and completeness) and model capacity.
* **Underfitting**: Occurs when a model is too simple (e.g., fitting a straight line to a complex curve) to capture the true underlying relationship between inputs and outputs.
* **Overfitting**: Occurs when a model is overly expressive by memorizing random noise and atypical details of the training data rather than the true pattern, resulting in poor predictions on unseen data.

### 2.2 Notes

#### 2.2.1 Loss vs Cost vs Objective functions

1. **Loss Function (Point-wise Error)**
    - Scope: Single data point.
    - Definition: Measures how poorly the model predicts the output for one specific input instance $(x_i, y_i)$.
    - Example: The squared error for point $i$:
    $$L_i = (\hat{y}_i - y_i)^2$$
2. **Cost Function (Dataset-wide Total)**
    - Scope: Entire dataset / model-wide parameters.
    - Definition: The aggregated error across all $N$ data points, plus optional global terms (like regularization parameters) that are not tied to individual data points.
    - Example: Mean Squared Error plus L2 regularization (we will read this in upcoming chapters):
    $$J(\boldsymbol{\phi}) = \frac{1}{N} \sum_{i=1}^{N} (\hat{y}_i - y_i)^2 + \lambda \Vert{}\boldsymbol{\phi}\Vert{}^2$$
3. **Objective Function (The Universal Umbrella)**
    - Scope: Any optimization target.
    - Definition: The most general term for any mathematical function that an algorithm attempts to either minimize or maximize.
    - Relationship: Every loss and cost function is an objective function, but an objective function can also be something you maximize (like expected reward in reinforcement learning ).
Summary Equation
$$\underbrace{\text{Objective Function } J(\boldsymbol{\phi})}_{\text{Function being optimized}} = \underbrace{\frac{1}{N} \sum_{i=1}^{N} \overbrace{L_i(\hat{y}_i, y_i)}^{\text{Loss Function}}}_{\text{Dataset Average Loss}} + \underbrace{\text{Regularization Term}}_{\text{Non-data point cost term}}$$

Long story short, you can say that:
A loss function is a part of a cost function which is a type of objective function.
