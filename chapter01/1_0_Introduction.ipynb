# Chapter 01: INTRODUCTION

* **Artificial intelligence (AI):** Builds systems that simulate intelligent behavior using logic, search, and probabilistic reasoning.
* **Machine Learning (ML):** A subset of AI that fits mathematical models to observed data to make decisions; currently (and incorrectly) used almost interchangeably with "AI."
* **Deep Learning (DL):** The process of fitting deep neural networks, currently the most powerful and practical type of ML models—to data.
Some everyday deep learning applications include
  -  **Natural Language Processing:** Translating text across languages.
  - **Computer Vision:** Searching and identifying images of specific objects.
  - **Speech Recognition:** Interacting with digital voice assistants.
* **Three ML Domains:** It can coarsely be divided into three areas (all currently driven by deep learning):
  1. Supervised Learning
  2. Unsupervised Learning
  3. Reinforcement Learning

### **1. Supervised Learning**
Supervised (meaning with supervision) learning models define a mathematical mapping from input data to an output prediction.
*   **Regression vs. Classification:** 
    *   **Regression models** predict a continuous numerical value, such as a house price or molecular freezing and boiling points. It can be single or multi-variable based on no of variables it's predicting.
    *   **Classification models** assign inputs to categorise and return category probabilities. This includes **binary classification** - mapping to one of two categories, like positive/negative reviews and **multiclass classification** - assigning to $N > 2$ categories, like classifying audio genres or objects in an image.

*   **Inputs:** In supervised learning, inputs vary from **tabular data** (which has no internal structure, meaning reordering variables does not change predictions) to complex structured inputs like text, audio, images, or molecular geometries. For example 
    - (a) input is a tabular data with fixed length vector which characterizes the property 
    - (b) input molecule may contain varying no of atoms connected in different way, so model must consider both geometric structure of molecule and the constituent atoms to the model ([RDKit](https://www.rdkit.org/docs/GettingStartedInPython.html), [SMILES](https://en.wikipedia.org/wiki/Simplified_Molecular_Input_Line_Entry_System))
    - (c) input is a text which can be of variable length depending on the number of words in the review and also the input order is important, *my wife ate the chicken* is not the same as *the chicken ate my wife.*
    - (d) input may be 10 sec clip which is very high-dimensional, for a 10 sec digital audio sampled at 44.1 kHz and represented in 16-bit integers, it will contain 441,000 integers.
    - (e) input is a concatenated RGB values at every pixel and also need to consider spatial structure (position of each pixel wrt other or certain pixels.)

*   **Training and Fitting:** A machine learning model represents an entire family of equations mapping input to
the output (i.e., a family of different cyan curves). **Training** (or **fitting**) is the process of using a labeled training dataset of input/output pairs to search through these family equations which map inputs to correct outputs as closely as possible.

*   **Deep Neural Networks:** These are highly powerful equations representing a broad family of relationships that can process inputs which are very large, of variable length, and contain various kinds of internal structures.
They can output single real numbers (regression), multiple numbers (multivariate regression), or probabilities over two or more classes (binary and multiclass classification, respectively). In next section we will see, their outputs may also be very large, of variable length, and contain internal
structure.
*   **Structured Outputs:** Some supervised tasks produce high-dimensional, structured outputs (such as semantic segmentation, depth estimation, translation, transcription, or image synthesis). These outputs are more difficult because they have complex internal structures or "grammars". This grammar can be learned without labels using unsupervised models.

### **2. Unsupervised Learning**
Unsupervised (meaning without supervision) learning models are learned from observed data in the absence of output labels. Rather than learning an input-to-output mapping, the goal is to describe or understand the underlying structure of the data. Here also, the input data may have very different characteristics; it may be discrete or continuous, low or high-dimensional, and of constant or variable length.

*   **Generative Models:** These models learn to synthesize new, highly plausible data examples (such as realistic images or text) that are statistically indistinguishable from the training data (Fig 1.5 and 1.6). 
**Conditional Generation:** This involves synthesizing new data under the constraint that some outputs are predetermined; e.g., *Fig 1.7 image inpainting*, where obscured regions are reconstructed while keeping surrounding pixels unchanged, or *Fig 1.8 text completion*.

*   **Latent Variables:** Many generative models assume real-world data can be described by a smaller number of underlying *latent variables*. These latent variables are designed to have a simple probability distribution. Deep learning is used to describe the mapping between this low-dimensional latent space and the high-dimensional observed data space. For example, a generative model can take a photo of a dog and a photo of a cat, find their underlying latent variables, and blend them step-by-step to create realistic intermediate images that smoothly morph the dog into the cat, Figure 1.11.

* **Connecting Supervised & Unsupervised Learning** Generative models boost structured supervised learning (e.g., text-to-image) by mapping the text's latent variables to the image's latent variables, rather than mapping raw text directly to pixels, Figure 1.12. Key advantages are:
  - Efficiency: Operates in lower dimensions, requiring fewer text/image training pairs.
  - Plausibility: Latent variables naturally decode into realistic, valid images.
  - Diversity: Randomness in the latent space yields multiple distinct outputs for a single prompt.

### **3. Reinforcement Learning**
Reinforcement learning is a sequential decision-making framework where an *agent* which performs certain *actions* at each *time steps* in an *environment* to *maximize* received *cumulative rewards*. Its core concepts and challenges includes:

* **Dynamic & Non-Deterministic States:** Actions change the state of the world, but often in unpredictable ways (e.g., due to opponent responses or environmental factors).

* **Temporal Credit Assignment:** One complication is that rewards are often sparse and delayed, occurring some time after the specific action that caused them was taken. 

* **Exploration-Exploitation Trade-Off:** The agent must balance exploiting known strategies that yield rewards with exploring untested actions to discover better outcomes.

* **Policy Networks:** RL incorporates deep learning by using policy networks to map a given state directly to the best action.

##### 1. Humanoid Robot Locomotion
* **Actions & States:** Moving various joints alters the robot's physical pose and spatial orientation.
* **Temporal Credit Assignment:** Rewards are granted upon reaching checkpoints in an obstacle course. Because reaching a checkpoint requires many joint adjustments, isolating which exact movements contributed to success is difficult.
* **Exploration vs. Exploitation:** The robot must decide whether to exploit a known, sub-optimal method (e.g., sliding awkwardly on its side) or explore new joint movements to master efficient upright walking.
* **Policy Network Role:** Maps raw sensor measurements directly to joint movements.

##### 2. Playing Chess
* **Actions & States:** Making legal chess moves changes the board state non-deterministically, as the opponent can respond with various counter-moves.
* **Temporal Credit Assignment:** Because rewards are provided only at the end of the match for winning or losing, credit assignment becomes extreme—the system must determine which specific actions among dozens of prior moves actually caused the ultimate victory or defeat.
* **Exploration vs. Exploitation:** The agent must choose between sticking to a known, decent sequence of opening moves (exploitation) or trying entirely new opening variations (exploration).
* **Policy Network Role:** Maps the current visual/data state of the board directly to the selection of the best move.

### **4. AI Ethics**
While artificial intelligence offers enormous benefits across fields like healthcare and commerce, it carries equal potential for harm. Scientists, engineers, and researchers must critically consider five primary ethical concerns:

* **Bias and Fairness**: AI models trained on historical data tend to reproduce or worsen existing societal biases (e.g., wage disparities between genders or racial biases in facial processing).

* **Explainability**: Deep learning models operate as "black boxes," making complex decisions without clear explanations. While local explanations exist for specific outputs, overall system transparency remains unsolved.

* **Weaponization**: AI is actively being deployed in military applications, turning it into one of the most powerful and potentially dangerous technologies for war.

* **Concentrated Power & Job Displacement**: Controlled primarily by major corporations for profit, AI threatens to concentrate power among a few organizations while automating jobs and disproportionately impacting low-wage workers.

* **Existential Risk**: Highly capable and extensible AI systems pose existential threats to humanity, ranging from concentrated control by elite owners to a total loss of human control over AI behavior.

**Additional Risks & Responsibility**
Beyond these five areas, AI enables mass surveillance, disinformation/fake news, spam, privacy violations, political manipulation, financial manipulation, and revenge porn.

As corporate profits and rapid AI deployment move far faster than government regulations, AI professionals cannot simply rely on laws for protection. Instead, researchers, students, and engineers must hold themselves accountable by closely examining whether their organizations practice genuine safety or merely use "ethics-washing" for public relations, while actively advocating for real ethical standards and education.
